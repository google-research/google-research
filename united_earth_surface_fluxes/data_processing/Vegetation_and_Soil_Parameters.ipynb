For **Ice caps and glaciers (Row 11)**, **Inland water (Row 13)**, **Ocean (Row 14)**, and **Water and land mixtures (Row 19)**, the missing values (NaNs) are due to these surfaces *not being parameterized as active vegetation*.

| Parameter | Recommended Imputation Value | Physical Rationale (IFS Model Context) |
| :--- | :--- | :--- |
| **$H/L \text{ veg}$** (High/Low veg) | **$0$** | Consistent with the existing value for **Ocean** and other non-vegetated types like **Desert**. Represents low or no canopy height. |
| **$r_s, \text{min}$** (Min Stomatal Resistance) | **A very large constant** (e.g., $10^6 \ s \cdot m^{-1}$) | These surfaces have no stomata (no transpiration). This value is used in LSMs to effectively set the **evapotranspiration flux to zero** for the vegetation component of the surface. |
| **$c_{veg}$** (Vegetation Coverage) | **$0.00$** | No vegetation fraction, as seen in the existing value for **Ocean** and **Desert** is $0.00$. This is the clearest and safest imputation. |
| **$g_D$** (Decoupling Factor) | **$0.00$** | Related to vegetation's control over water flux; set to zero for non-vegetated surfaces where the formula is irrelevant. |
| **$a_r$ and $b_r$** (Root Distribution Attenuation) | **$0.000$ for both** | These coefficients define the **exponential root profile** ($f(z) \propto e^{-a_r z} / b_r$). For non-vegetated surfaces, there are **no roots**, so the entire root distribution function must be zero, achieved by setting these coefficients to $0$. This aligns with the existing $0.000$ values for **Water and land mixtures** (Row 19).  |

### Imputation for Open Water/Deep Water


| Parameter | Recommended Imputation Value | Physical Rationale |
| :--- | :--- | :--- |
| **$\theta_{\text{sat}}$ (Saturated Water Content)** | **$1.000$** | This is the volume fraction of water when saturated. In pure water, the volume of water equals the total volume of the medium. Some models use a value slightly less than 1 (e.g., $0.999$) to avoid division by zero errors in soil porosity calculations, but $1.000$ represents the physics best. |
| **$\theta_{\text{cap}}$ (Field Capacity)** | **$1.000$** | Field capacity is the water content after gravitational drainage. For open water, there is no matrix potential that would cause water to drain from the bulk liquid (unless it's draining out of the grid box, which is handled by a runoff or river flow scheme, not this parameter). Thus, the "holding capacity" is the same as the saturation point. |
| **$\theta_{\text{pwp}}$ (Permanent Wilting Point)** | **$0.000$** | This is the most crucial parameter. Wilting Point defines the minimum water content *available* for transfer. Setting this to zero means **all** the water in the layer ($\theta_{\text{sat}} - \theta_{\text{res}}$) is considered **plant-available water** (PAW). This is consistent with water being readily available for evaporation/sublimation. |
| **$\theta_{\text{res}}$ (Residual Water Content)** | **$0.000$** | This represents water held so tightly by the soil matrix that it can't be removed, even by plants. Since open water has no solid matrix, the residual water should be zero. |
| **$\theta_{\text{cap}} - \theta_{\text{pwp}}$ (Plant Available Water)** | **$1.000$** | This column is derived as $\theta_{\text{cap}} - \theta_{\text{pwp}}$, and using the imputed values yields $1.000 - 0.000 = 1.000$. This confirms that all the water in the layer is available for flux. |

| :--- | :--- | :--- |

### Ground heat flux has its own soil parameters, different from table_8_8_df. It is defined below

```python
data = {
    'Texture Class': [
        'Coarse (Sand)',
        'Medium (Loam)',
        'Medium-Fine',
        'Fine (Clay)',
        'Very Fine (Clay)',
        'Organic',
        'Trop. Organic',
    ],
    # K_dry [W m⁻¹K⁻¹]
    'K_dry [W m⁻¹K⁻¹]': [0.29, 0.24, 0.20, 0.16, 0.12, 0.05, 0.24],
    # K_sat [W m⁻¹K⁻¹]
    'K_sat [W m⁻¹K⁻¹]': [2.50, 2.05, 1.83, 1.58, 1.33, 0.50, 2.00],
    # C_dry [10⁶ J m⁻³K⁻¹]
    'C_dry [10⁶ J m⁻³K⁻¹]': [1.28, 1.38, 1.48, 1.58, 1.68, 0.51, 1.38],
    # θ_sat [m³/m³] (using the Theta character 'θ')
    'θ_sat [m³/m³]': [0.395, 0.435, 0.485, 0.482, 0.482, 0.835, 0.439],
}

ground_heat_soil_params = pd.DataFrame(data)

display(ground_heat_soil_params)
```



### Values to Define Water's Properties



| Parameter | Imputation Value for Open Water | Physical Basis |
| :--- | :--- | :--- |
| **$\text{K}_{\text{dry}}$** [$\text{W m}^{-1} \text{K}^{-1}$] | **$\mathbf{0.0}$** | Water has no "dry" state (no solid matrix). |
| **$\text{K}_{\text{sat}}$** [$\text{W m}^{-1} \text{K}^{-1}$] | **$\approx \mathbf{0.60}$** | This is the thermal conductivity of liquid water (e.g., $0.60 \text{ W/m} \cdot \text{K}$ at $20^\circ \text{C}$). |
| **$\text{C}_{\text{dry}}$** [$10^6 \text{ J m}^{-3} \text{K}^{-1}$] | **$\approx \mathbf{4.18}$** | This is the volumetric heat capacity of water (which is its specific heat capacity $\times$ density). |
| **$\theta_{\text{sat}}$** [$\text{m}^3/\text{m}^3$] | **$\mathbf{1.00}$** | Open water is 100% water by volume. |


```python
new_C_dry_value = (
    ground_heat_soil_params['C_dry [10⁶ J m⁻³K⁻¹]'].max() * 2
)  # Instead of actual value, just to avoid big numbers for ML training.
water_gh_row_data = {
    'Texture Class': ['Water'],
    'K_dry [W m⁻¹K⁻¹]': [0.0],
    'K_sat [W m⁻¹K⁻¹]': [0.6],
    'C_dry [10⁶ J m⁻³K⁻¹]': [new_C_dry_value],
    'θ_sat [m³/m³]': [1.0],
}
water_gh_row_df = pd.DataFrame(water_gh_row_data)
ground_heat_soil_params = pd.concat(
    [water_gh_row_df, ground_heat_soil_params], ignore_index=True
)
print("DataFrame for Ground Heat Soil Parameters with 'Water' row:")
display(ground_heat_soil_params)
```

```python
ground_heat_soil_params = ground_heat_soil_params.drop(
    columns=['Texture Class']
)
display(ground_heat_soil_params)
```

```python
print('Before normalizing')
display(ground_heat_soil_params)
ground_heat_soil_params_normalized = (
    ground_heat_soil_params / ground_heat_soil_params.iloc[1:].max(0)
)
print('\nAfter normalizing')
display(ground_heat_soil_params_normalized)
```
